# Rivers and Stream Power Law

Niagara falls are advecting over a relatively short river segment where discharge can be considered constant. This is not the case for all rivers where discharge typically increases as a power function of distance downstream. Let's simulate how bedrock rivers evolve through time and they respond to changes in climate and tectonics, and for that we will use a detachment-limited equation.

The detachment-limited equation is as follows: Consider a location on a stream channel that has local downstream slope gradient $S$ and drainage area $A$. We define an erosion $E$ as:

$$
E = KA^m S^n \tag{1}
$$

where $K$ is an erodibility coefficient with dimensions of $[L^{(1-2m)}/T]$. $K$ is thought to be positively correlated with climate wetness, or storminess (this is hard to quantify) and to be negatively correlated with rock strength (again, rock strength is hard to quantify). $A$ is drainage area and $S$ is the slope. The erosion function has dimensions of erosion (lowering) rate, [L/T]. The expression is also known as the "stream power law" because the exponents can be configured to represent an erosion law that depends on stream power per unit bed area (Whipple & Tucker, 1999). A common choice of exponents is $m = 1/2$, $n=1$ but other combinations are possible depending on one's assumptions about process, hydrology, channel geometry, and other factors (e.g., Howard et al., 1994; Whipple et al., 2000).

Rewriting Fluvial incision as a change of elevation in time, and $S$ as the slope of the steepest descent $(-\frac{dz}{dx})$ where $x$ is horizontal distance (positive in the downslope direction) and $z$ is elevation results in the following partial differential equation:

$$
\frac{\partial z}{\partial t} = -K A^m (\frac{\partial z}{\partial x}) ^n \tag{2}
$$

Now we want to understand how changing climate, lithology and tectonics would change fluvial incision and landscape evolution. For that we will use the advection solver function. Please run cell below:  

````python
import numpy as np
from matplotlib import pyplot as plt

def stream_power_solver(T_num,K,dx,run_duration,U, m= 0.5,n=1):
    v=-K*A**m    
    dt = dx/np.max(np.abs(v))*0.9
    print('dt is %.02f year'%dt)
    iter = 0
    #Numerical solution 
    while iter * dt < run_duration:
        
        iter += 1
        if v[-1] > 0:
            T_num[1:] -= v[1:] * dt  * ((T_num[1:]-T_num[:-1])/ dx)**n
            T_num[1:] += U*dt
        elif v[-1] < 0:            
            T_num[:-1] -= v[:-1] * dt  * ((T_num[1:]-T_num[:-1])/ dx)**n
            T_num[:-1] += U*dt
    return T_num
````

Now that we have a function that numerically solves the stream power equation, let's try to represent the evolution of Niagara falls in a 1D river profile! We will simulate a river of 15 km, using spatial steps ($dx$) of 100m, where the river is at 70m elevation except for base level, sitting at 0 m. 
We also assume that discharge ($A$) is a power function of distance with an exponent of 1.5 ($A = x^{1.5}$). It would be unrealistic to have zero discharge at the upmost cell. To correct for this, represent discharge as $A=x^{1.5}+dx$. After defining our spatial domain, river elevation and drainage area, we will plot the original topography and drainage area along the river profile. 

Let's go though this together:

````python
L = 15000 
nb_cells = 151 
dx = L / (nb_cells-1)
print('dx is %i m'%dx)

x=np.linspace(0,L,nb_cells)
z=np.ones(nb_cells)*70
z[-1]= 0
z_ini = np.array(z)
A = np.power(x,1.5)+dx
plt.plot(x,A)
plt.xlabel('Distance, m')
plt.ylabel('Drainage area, m$^2$')
plt.show()

plt.plot(x,z_ini,label =  'Initial surface', color = 'green')
plt.xlabel('Distance, m')
plt.ylabel('Elevation, m')
plt.legend(loc='lower left')
plt.show()
````

Now let's run the stream power solver, without any uplift for now, plotting the original river profile and the river profile after $10^4$ years. 

````python

run_duration = 10000
U = 0
K = 5*1e-4
z=np.ones(nb_cells)*70
z[-1]= 0
z_ini = np.array(z)
z = stream_power_solver(z,K,dx,run_duration,U, m=0.5,n=1)
plt.plot(x,z_ini,label =  'Initial surface', color = 'green')
plt.plot(x,z,label =  'Solution, 10 kyr', color = 'black')
plt.xlabel('Distance, m')
plt.ylabel('Elevation, m')
plt.legend(loc='lower left')
plt.show()
````

$$

where $(\frac{U}{K})^\frac{1}{n}$ is the steepness index $k_s$, and $-\frac{m}{n}$ is the concavity index $\theta$
